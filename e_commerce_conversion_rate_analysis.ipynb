```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("dataset_files/online_shoppers_intention.csv")

print(df.head())
```

```python
plt.figure(figsize=(6,4))

sns.countplot(x='Revenue', data=df)

plt.title('Revenue Distribution')
plt.xlabel('Revenue')
plt.ylabel('Number of Visitors')

plt.show()
```

```python
print(df["Revenue"].value_counts())
```

```python
print(df["Revenue"].value_counts(normalize=True) * 100)
```

```python
total_sessions = len(df)

converted_sessions = df["Revenue"].sum()

conversion_rate = (converted_sessions / total_sessions) * 100

print("Total Sessions:", total_sessions)
print("Converted Sessions:", converted_sessions)
print("Conversion Rate:", round(conversion_rate, 2), "%")
```
